#### Extraccion del excel de encuesta de preferencias a formato lista-diccionario

```python
def eliminar_parentesis(texto):
    resultado = ""
    nivel = 0

    for c in texto:
        if c == '(':
            nivel += 1
        elif c == ')':
            if nivel > 0:
                nivel -= 1
        else:
            if nivel == 0:
                resultado += c

    return resultado


def separar_valores(texto):
    if pd.isna(texto):
        return []

    texto = str(texto)

    limpio = eliminar_parentesis(texto)

    partes = limpio.split(',')

    return [p.strip() for p in partes if p.strip()]

df = pd.read_excel(
    r"C:\Users\gtoma\Desktop\Encuesta_preferencias.xlsx",
    usecols="B:E"
)

resultado = []

for _, row in df.iterrows():
    fila_dict = {}

    for col in df.columns:
        fila_dict[col] = separar_valores(row[col])

    resultado.append(fila_dict)

for i, r in enumerate(resultado):
    print(f"Fila {i+1}: {r}")
```

#### Extracion de horas de visualizacion y en que se fija la gente (CRITERIO 1)

```python
criterio1 = []
for i, r in enumerate(resultado):
    new_dicc = {}
    new_dicc['¿En qué te fijas a la hora de ver un anime?'] = r['¿En qué te fijas a la hora de ver un anime?']
    new_dicc['¿Con qué frecuencia consumes Anime a la semana?'] = r['¿Con qué frecuencia consumes Anime a la semana?']
    criterio1.append(new_dicc)

# Mostrar resultados
for i, r in enumerate(criterio1):
    print(f"Fila {i+1}: {r}")
```

#### Extracion de como te enteras de cada anime y en que se fija la gente (CRITERIO 2)

```python
criterio2 = []
for i, r in enumerate(resultado):
    new_dicc = {}
    new_dicc['¿En qué te fijas a la hora de ver un anime?'] = r['¿En qué te fijas a la hora de ver un anime?']
    new_dicc['¿Cómo te enteras de nuevos Animes?'] = r['¿Cómo te enteras de nuevos Animes?']
    criterio2.append(new_dicc)

# Mostrar resultados
for i, r in enumerate(criterio2):
    print(f"Fila {i+1}: {r}")
```

#### Visualización en Ridgeline Plot para el Criterio 1

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

mapping = {
    "0-4 horas": "0-4",
    "4-8 horas": "4-8",
    "8-16 horas": "8-16",
    "+16 horas": "16+"
}

rows = []

for fila in criterio1:

    criterios = fila['¿En qué te fijas a la hora de ver un anime?']
    frecuencias = fila['¿Con qué frecuencia consumes Anime a la semana?']

    for criterio in criterios:
        for frecuencia in frecuencias:

            if frecuencia in mapping:

                rows.append({
                    "criterio": criterio,
                    "horas": mapping[frecuencia]
                })

df = pd.DataFrame(rows)

replace_dict = {
    "A través de recomendación": "Recomendación",
    "Publicidad física o en servicios no relacionados a los anteriores": "Publicidad",
    "Plataformas de Catalogación y Rankings": "Rankings",
    "Banda sonora/OST": "OST",
    "Estudio de animación": "Estudio",
}

df["criterio"] = df["criterio"].replace(replace_dict)

top_criterios = (
    df['criterio']
    .value_counts()
    .head(8)
    .index
)

df = df[df['criterio'].isin(top_criterios)]

frecuencia_total = (
    df['criterio']
    .value_counts()
)

tabla = pd.crosstab(
    df['criterio'],
    df['horas'],
    normalize='index'
)

column_order = ["0-4", "4-8", "8-16", "16+"]

for col in column_order:
    if col not in tabla.columns:
        tabla[col] = 0

tabla = tabla[column_order]

max_freq = frecuencia_total.max()

scales = (
    frecuencia_total / max_freq
).values

N = len(tabla)

angles = np.linspace(0, 2*np.pi, N, endpoint=False)

width = 2*np.pi / N * 0.72

colors = [
    "#A8DADC",  # celeste suave
    "#457B9D",  # azul acero
    "#1D3557",  # navy
    "#E63946"   # rojo intenso
]

fig, ax = plt.subplots(
    figsize=(14,14),
    subplot_kw=dict(polar=True)
)

# Fondo limpio
fig.patch.set_facecolor("white")
ax.set_facecolor("white")

bottom = np.zeros(N)

for i, categoria in enumerate(column_order):

    values = tabla[categoria].values * scales

    ax.bar(
        angles,
        values,
        width=width,
        bottom=bottom,
        color=colors[i],
        edgecolor="white",
        linewidth=2.5,
        alpha=0.98,
        label=categoria
    )

    bottom += values

ax.set_xticks(angles)

ax.set_xticklabels(
    tabla.index,
    fontsize=12,
    fontweight='bold'
)

ax.tick_params(
    axis='x',
    pad=30
)

ax.grid(False)

ax.set_yticks([])

ax.spines['polar'].set_visible(False)

plt.title(
    "Preferencias de Anime y Distribución de Consumo Semanal",
    fontsize=20,
    pad=50,
    fontweight='bold'
)

legend = plt.legend(
    title="Horas semanales",
    bbox_to_anchor=(1.18, 1.08),
    frameon=False,
    fontsize=11
)

legend.get_title().set_fontsize(12)

plt.tight_layout()
plt.show()
```

#### Visualización en Chord diagram para el Criterio 2

- **Instalar holoviews para visualizar el Chord Diagram**: conda install -c pyviz holoviews bokeh
- **Instalar pycirclize para ver las etiquetas ordenadas**: conda install -c conda-forge pycirclize

```python
from collections import Counter
from pycirclize import Circos
from pycirclize.parser import Matrix
import pandas as pd
import matplotlib.pyplot as plt

connections = []

replace_dict = {
    "A través de recomendación": "Recomendación",
    "Publicidad física o en servicios no relacionados a los anteriores": "Publicidad",
    "Plataformas de Catalogación y Rankings": "Rankings",
    "Plataformas de Streaming": "Streaming",
    "Banda sonora/OST": "OST",
    "Estudio de animación": "Estudio",
    "Duración / Capítulos": "Duración",
}

for fila in criterio2:

    criterios = fila['¿En qué te fijas a la hora de ver un anime?']
    medios = fila['¿Cómo te enteras de nuevos Animes?']

    # Aplicar reemplazos
    criterios = [replace_dict.get(x, x) for x in criterios]
    medios = [replace_dict.get(x, x) for x in medios]

    for c in criterios:
        for m in medios:
            connections.append((c, m))

counter = Counter(connections)

df = pd.DataFrame([
    [source, target, value]
    for (source, target), value in counter.items()
], columns=["from", "to", "value"])

df = df[df["value"] >= 2]

matrix = Matrix.parse_fromto_table(df)

criterios_order = sorted(df["from"].unique())
medios_order = sorted(df["to"].unique())

sector_order = criterios_order + medios_order


circos = Circos.initialize_from_matrix(
    matrix,
    start=-90,
    end=270,
    space=12,
    r_lim=(90, 100),
    cmap="tab20",
    order=sector_order,
    label_kws=dict(size=9)
)

fig = circos.plotfig(figsize=(14, 14))

plt.title(
    "Relación entre Preferencias y Descubrimiento de Anime",
    fontsize=18,
    pad=30
)

plt.show()
```
