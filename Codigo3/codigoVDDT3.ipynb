### Tratamiento de datos para frecuencia de usuarios/pais MyAnimeList

```python
UserList = pd.read_csv("UserList.csv")
print(UserList.head())
print(UserList.shape)
```

```python
UL_location = UserList[['location']].dropna().copy()
```

```python
print(UL_location.head())
print(UL_location.value_counts())
```

```python
import pycountry

country_names = {
    country.name.lower(): country.name
    for country in pycountry.countries
}
```

```python
import pandas as pd
import pycountry

# Diccionario de países
country_names = {
    country.name.lower(): country.name
    for country in pycountry.countries
}

def extraer_pais(texto):

    texto = str(texto).lower()

    for nombre, oficial in country_names.items():
        if nombre in texto:
            return oficial

    return None

# Crear columna país
UL_location["country"] = UL_location["location"].apply(extraer_pais)

# Contar países encontrados
country_df = (
    UL_location["country"]
    .dropna()
    .value_counts()
    .reset_index()
)

country_df.columns = ["country", "count"]
```

```python
print(country_df.size)
```

```python
population = 0
for i in country_df["count"]:
    population += i
print(population)
```

```python
country_df.to_csv(
    "anime_users_by_country.csv",
    index=False,
    encoding="utf-8"
)

print("Archivo exportado")
```
